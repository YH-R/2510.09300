## SiC Drude-Lorentz model

- Incandescent (Main): Page 4

\begin{align*}
    \epsilon_{\textrm{eq}}(\omega) &= \epsilon_\infty \frac{\omega_L^2 - \omega^2 - i\gamma \omega}{\omega_T^2 - \omega^2 - i \gamma \omega}
\end{align*}

```python
# Parameters for the Drude-Lorentz model
wL = 29.1 * (2*pi*THz)
wT = 23.8 * (2*pi*THz)
damp = 0.14 * (2*pi*THz)
eps_inf = 6.7

def permittivity(w):
    return eps_inf * (wL**2 - w**2 - 1j * damp * w) / (wT**2 - w**2 - 1j * damp * w)
```

```python
def plot_eps():
    w_arr = np.linspace(w_MIN, w_MAX, 290)
    eps_arr = np.array([permittivity(w) for w in w_arr])
    f_arr = w_arr / (2*pi*THz) # Display as frequency in THz
    
    plt.figure(figsize=(4, 3))
    
    plt.plot(f_arr, eps_arr.real, label=rf"Re($\epsilon$)")
    plt.plot(f_arr, eps_arr.imag, label=rf"Im($\epsilon$)")
    
    plt.ylim(-20, 20) # uncomment to see full scale
    plt.title("Drude-Lorentz Permittivity")
    plt.xlabel(r"$\omega/2\pi$ [THz]")
    plt.ylabel(r"$\epsilon$")
    plt.legend()
    plt.show()
    
plot_eps()
```

## Complex wavevectors

**Notations in Incandescent** (SM Page 13)

$\vec{k}_j = k_0 (\kappa_x, \kappa_y, \sqrt{\epsilon_j} \kappa_{z,j})$

$\kappa_R = \sqrt{\kappa_x^2 + \kappa_y^2}$

$\kappa_{z,j} = \sqrt{1 - \kappa_R^2/\epsilon_j}$

**Notations in Surface**

$\vec{k}_j = (k_x, k_y, \gamma_j)$

$|\boldsymbol{K}| = k_\perp = \sqrt{k_x^2 + k_y^2} = k_0 \kappa_R$

$\gamma_j = \sqrt{\epsilon_j \frac{\omega^2}{c^2} - k_\perp^2} = k_0 \sqrt{\epsilon_j} \kappa_{z,j} = k_{z,j}$

How to choose sign of square root?
- $\Im(\tilde{k}_1) > 0$: For the expression $e^{-2 k_0 \Im(\tilde{k}_1) z}$ to decay as $z \to \infty$.
- $\Im(\tilde{k}_2) > 0$: For $\mathcal{S}_{0}$ to be positive (Eq. S.VI.25).

```python
import cmath
my_sqrt = cmath.sqrt # principal value, can try other definitions

def kappa_z(kR, eps):
    return my_sqrt(1+0j - kR**2/eps)

def k_tilde(kR, eps):
    return my_sqrt(eps - kR**2)
```

### Floquet-shifted wavevector

- Incandescent (SM): Page 20

\begin{align*}
    \gamma_i^{\pm} &= \sqrt{\epsilon_i^\pm \frac{(\omega^\pm)^2}{c^2} - k_\perp^2} \\
    \tilde{k}_{i}^\pm &= \sqrt{\epsilon_i^\pm (1 \pm \tilde{\Omega})^2 - \kappa_R^2} \\
    \kappa_{i}^\pm &= \sqrt{1 - \kappa_R^2/ [(1 \pm \tilde{\Omega})^2 \epsilon_i(\omega^\pm)]} 
\end{align*}

where
\begin{align*}
    \tilde{\Omega} &= \Omega/\omega \\
    \omega^\pm &= \omega \pm \Omega \\
    \epsilon_i^\pm &= \epsilon_i(\omega^\pm)
\end{align*}

```python
def kappa_z_pm(kR, eps_pm, one_tilde):
    return my_sqrt(1+0j - kR**2/(one_tilde**2 * eps_pm))

def k_tilde_pm(kR, eps_pm, one_tilde):
    return my_sqrt(eps_pm * one_tilde**2 - kR**2)
```

## Fresnel Coefficients

### Transmission coefficients

- Incandescent: Eqs. (S.VI.8a) and (S.VI.8b)
- Surface: Eq. (A.5)

\begin{align*}
        t^s_{i \leftarrow j} &= \frac{2 \sqrt{\epsilon_j} \kappa_j}{\sqrt{\epsilon_i} \kappa_i + \sqrt{\epsilon_j} \kappa_j} 
        = \frac{2 \gamma_j}{\gamma_i + \gamma_j} \\
        t^p_{i \leftarrow j} &= \frac{2 \sqrt{\epsilon_j} \kappa_j}{\sqrt{\epsilon_j} \kappa_i + \sqrt{\epsilon_i} \kappa_j} 
        = \frac{2 \sqrt{\epsilon_i \epsilon_j} \gamma_j}{\epsilon_j \gamma_i + \epsilon_i \gamma_j} 
\end{align*}

Notation
- Incandescent: $t^s_{ji}$
- Surface: $t^s_{i \leftarrow j}$

They depend on $\omega$ and $k_\perp$, but not on the direction of $\vec{k}_\perp$. 

Note that $t^p(-\omega) = t^p(\omega)^*$

```python
def ts(k_til_i, k_til_j):
    return 2*k_til_j/(k_til_i + k_til_j)

def tp(k_til_i, k_til_j, eps):
    return 2*my_sqrt(eps)*k_til_j/(eps*k_til_i + k_til_j)
```

### Reflection coefficients

- Incandescent: Eqs. (S.VI.11a) and (S.VI.11b)
- Surface: Eq. (A.8)

\begin{align*}
        r^s_{j \leftarrow i} &= \frac{\sqrt{\epsilon_j} \kappa_j - \sqrt{\epsilon_i} \kappa_i}{\sqrt{\epsilon_i} \kappa_i + \sqrt{\epsilon_j} \kappa_j} 
        = \frac{\gamma_j - \gamma_i}{\gamma_i + \gamma_j} \\
        r^p_{j \leftarrow i} &= \frac{\sqrt{\epsilon_i} \kappa_j - \sqrt{\epsilon_j} \kappa_i}{\sqrt{\epsilon_j} \kappa_i + \sqrt{\epsilon_i} \kappa_j} 
        = \frac{\epsilon_i \gamma_j - \epsilon_j \gamma_i}{\epsilon_j \gamma_i + \epsilon_i \gamma_j} 
\end{align*}

Notation
- Incandescent: $r^s_{ij}$
- Surface: $r^s_{j \leftarrow i} = - r^s_{i \leftarrow j}$

Note: i and j are flipped compared to transmission coefficients, in both Incandescent and Surface. Furthermore, Incandescent and Surface definitions seem to differ by a minus sign. Here, we will follow Incandescent. 

```python
def rs(k_til_i, k_til_j):
    return (k_til_j - k_til_i)/(k_til_i + k_til_j)

def rp(k_til_i, k_til_j, eps):
    return (k_til_j - eps*k_til_i)/(eps*k_til_i + k_til_j)
```

## Parameters dictionary

Use a dictionary to store all the variables/parameters, this reduces any repeated calculations. Each time a different $\omega$ is used, the parameters need to be calculated again.

```python
def get_params(kR, w, z, Omega=None, sign=None): 
    # these need to be calculated once per integration point
    # Use Omega=None if we only want S0
    
    d = {'kR':kR, 'w':w, 'z':z}
    
    d['kR_sq'] = kR**2                     # standardized naming, "_sq" suffix means absolute squared
    d['eps'] = permittivity(w)
    d['n'] = my_sqrt(d['eps'])
    d['k0'] = w / c_speed
    d['k1'] = kappa_z(kR, 1)               # kappa_1
    d['k2'] = kappa_z(kR, d['eps'])        # kappa_2
    d['k2_til'] = k_tilde(kR, d['eps'])    # k1_til = k1
    
    d['decay'] = np.exp(-2 * d['k0'] * d['k1'].imag * z)
    
    d['ts'] = ts(d['k1'], d['k2_til'])
    d['tp'] = tp(d['k1'], d['k2_til'], d['eps'])
    d['ts_sq'] = abs(d['ts'])**2
    d['tp_sq'] = abs(d['tp'])**2
    
    if not Omega is None: 
        # this block is not entered to save time if we only want to calculate S0
        # then, sign should be either +1 or -1
        
        d['Omega'] = Omega
        d['sign'] = sign
        d['one_tilde'] = 1 + sign * Omega / w
        
        d['eps_pm'] = permittivity(w + sign * Omega)
        
        d['k1_til_sq'] = abs(d['k1'])**2
        d['k2_til_sq'] = abs(d['k2_til'])**2
        d['k2_til_pm'] = k_tilde_pm(kR, d['eps_pm'] , d['one_tilde']) # k1_til_pm = k1_til
        
        d['T'] = S02_T(d['k1_til_sq'], d['k2_til_sq'], d['kR_sq'], d['eps'])
        d['R'] = S02_R(d['k1_til_sq'], d['k2_til_sq'], d['kR_sq'], d['eps'])
        
        d['rs'] = rs(d['k1'], d['k2_til'])
        d['rp'] = rp(d['k1'], d['k2_til'], d['eps'])
        d['rs_pm'] = rs(d['k1'], d['k2_til_pm'])
        d['rp_pm'] = rp(d['k1'], d['k2_til_pm'], d['eps_pm'])
    
        # used frequently in second order calculations
        d['k2k2'] = d['k2_til'] * d['k2_til_pm']
        d['k2k2*'] = d['k2_til'] * d['k2_til_pm'].conjugate()
        d['k2*k2'] = d['k2_til'].conjugate() * d['k2_til_pm']
        d['denom'] = d['one_tilde']**2 * d['eps'] * d['eps_pm']
        
        d['T_pm'] = S11_T(d['k1_til_sq'], abs(d['k2_til_pm'])**2, d['kR_sq'], d['eps_pm'], d['one_tilde'])
        d['R_pm'] = S11_R(d['k1_til_sq'], abs(d['k2_til_pm'])**2, d['kR_sq'], d['eps_pm'], d['one_tilde'])
    
    return d
```

Flexible function to plot integrand when integrating w.r.t. $\kappa_R$. Useful for debugging. Integrand function should follow the signature `<name>_integrand(params)`.

```python
def plot_integrand(name, z, Omega=None, sign=None):
    func_name = name + "_integrand"
    if not func_name in globals():
        print(f"Function '{func_name}' not found.")
        return None
    
    func = globals()[func_name] # retrieve function using name as string
    
    kR_MIN = 1e-6 # avoid 0
    kR_MAX = 2
    kR_arr = np.linspace(kR_MIN, kR_MAX, kR_len)
    w_arr = np.linspace(w_MIN, w_MAX, w_len)
    
    kR_mesh, w_mesh = np.meshgrid(kR_arr, w_arr)
    
    params_mesh = [[None for j in range(kR_len)] for i in range(w_len)]
    for i in range(w_len):
        w = w_arr[i]
        for j in range(kR_len):
            kR = kR_arr[j]
            params_mesh[i][j] = get_params(kR, w, z, Omega=Omega, sign=sign)

    integrand_mesh = np.array([[func(d) for d in row] for row in params_mesh])
    max_val = np.max(np.abs(integrand_mesh))
    
    plt.figure(figsize=(5, 5))
    plt.imshow(integrand_mesh, 
               vmax=-max_val, 
               vmin=max_val, # set white as zero
               extent=[kR_MIN, kR_MAX, w_MIN/(2*pi*THz), w_MAX/(2*pi*THz)], 
               origin='lower', 
               aspect = (kR_MAX-kR_MIN)/(w_MAX-w_MIN)*(2*pi*THz), # make plot square
               cmap='bwr_r')
    
    plt.title(f"Heatmap of {name} integrand")
    plt.xlabel(r"$\kappa_R$")
    plt.ylabel(f"$f$ [THz]")
    plt.colorbar()
    plt.show()
```

## Zeroth-order term

### $\mathcal{S}_{0,0}(\omega)$ calculations

Refer to manuscript for the full theory, here the focus in on evaluating the integration of the Green's functions.

\begin{align*}
    \mathcal{S}_{0,0} &= 4 \epsilon_0 \hbar \omega^4 N(\omega) \Theta(\omega) \Im[\epsilon_{\textrm{eq}}(\omega)] \mu_0^2 \mathcal{G}_{0,0}(\vec{r}, \omega) \\
    \mathcal{G}_{0,0}(\vec{r}, \omega) &= \Re \operatorname{Tr} \int_{z' < 0} d^3 \vec{r}' \, G(\vec{r}, \vec{r}', \omega) G(\vec{r}', \vec{r}, \omega)^\dagger
\end{align*}

- Incandescent: Eqs. (S.VI.14) and (S.VI.24), differ by overall factor of $\pi$. 

```python
def S0(w, z):
    return 4 * eps0 * w**4 * bose(w) * permittivity(w).imag * mu0**2 * G00(w, z)
```

#### Computing the Green functions

- Incandescent (SM): Eq. (S.VI.24)

\begin{align*}
    \mathcal{G}_{0,0}(z, \omega) &= \frac{1}{k_0} \int_{0}^{\infty} \frac{\kappa_R d \kappa_R}{16 \pi} 
    \frac{1}{|\epsilon \kappa_2^2| \Im (\sqrt{\epsilon} \kappa_2)} \biggl[
        |t_s|^2 
        + |t_p|^2 (|\kappa_1|^2 + \kappa_R^2) (|\kappa_2|^2 + \frac{\kappa_R^2}{|\epsilon|})
    \biggr]
    e^{-2 k_0 \Im\kappa_1 z} 
\end{align*}

```python
# excluding factor of 1/(16 pi k_0)
def G00_integrand(params):
    G00s2 = params['ts_sq']
    G00p2 = params['tp_sq'] * (
        abs(params['k1'])**2 + params['kR_sq'] 
    ) * (
        abs(params['k2_til'])**2 + params['kR_sq'] 
    ) / abs(params['eps'])
    
    return (
        params['kR'] / (abs(params['k2_til'])**2 * params['k2_til'].imag) 
        * (G00s2 + G00p2) * params['decay']
    )

# when doing the integral using Scipy, we need to pass in the argument directly, not as part of a dictionary
def G00_integrand_wrapper(kR, w, z):
    params = get_params(kR, w, z)
    return G00_integrand(params)
```

```python
def G00(w, z):
    left_end = 0
    # right_end = 4
    right_end = np.infty
    
    integral, _ = integrate.quad(G00_integrand_wrapper, 
                                 left_end, 
                                 right_end, 
                                 args=(w, z))
    
    k0 = w/c_speed
    return integral/(16*pi*k0)
```

```python
def plot_S0(z, label="", load=False):
    w_arr = np.linspace(w_MIN, w_MAX, w_len)
    
    if load:
        # assume correct size
        S0_arr = np.load(label+'_S0.npy')
    else:
        S0_arr = np.array([S0(w, z) for w in w_arr])
        np.save(label+'_S0.npy', S0_arr)
    
    plt.figure(figsize=(6, 3))
    plt.plot(w_arr/(2*pi*THz), S0_arr*EE_FREQ_SI, label=rf"$z$={z/micron:.3g}um, $T$={temp}K")
    
    if label:
        title = f"S0 ({label})"
    else:
        title = f"S0"
    
    # plt.ylim(0, 8e-14)
    plt.title(title)
    plt.xlabel(r"$\omega/2\pi$ [THz]")
    plt.ylabel(r"$\mathcal{S}_0$ [$\mathrm{V}^2 \mathrm{m}^{-2} \mathrm{s}$]")
    # plt.xscale('log')
    plt.legend()
    plt.show()
```

## Second-order terms

### Projection factors

- Incandescent: Eqs. (S.VI.60a), (S.VI.60b)


\begin{align*}
    T &= \frac{1}{|\epsilon|} (|\tilde{k}_1|^2 + \kappa_R^2 ) (|\tilde{k}_2|^2 + \kappa_R^2 ) \\ 
    R &= \frac{1}{|\epsilon|} (|\tilde{k}_1|^2 + \kappa_R^2 ) (|\tilde{k}_2|^2 - \kappa_R^2) 
\end{align*}

Used in (0,2) calculations. Note that these are purely real quantities. 

```python
def S02_T(k1_til_sq, k2_til_sq, kR_sq, eps):
    return (k1_til_sq + kR_sq) * (k2_til_sq + kR_sq) / abs(eps)

def S02_R(k1_til_sq, k2_til_sq, kR_sq, eps):    
    return (k1_til_sq + kR_sq) * (k2_til_sq - kR_sq) / abs(eps)
```

- Incandescent: Eqs. (S.VI.41a), (S.VI.41b)

\begin{align*}
    T^\pm &= \frac{1}{(1 \pm \tilde{\Omega})^2 |\epsilon(\omega^\pm)|} 
        (|\tilde{k}_1|^2 + \kappa_R^2 ) (|\tilde{k}_2^\pm|^2 + \kappa_R^2 ) \\
    R^\pm &= \frac{1}{(1 \pm \tilde{\Omega})^2 |\epsilon(\omega^\pm)|} 
        (|\tilde{k}_1|^2 + \kappa_R^2 ) (|\tilde{k}_2^\pm|^2 - \kappa_R^2 ) 
\end{align*}

Only used in (1,1) calculations, but put here since the form is similar. 

```python
def S11_T(k1_til_sq, k2_til_pm_sq, kR_sq, eps_pm, one_tilde):
    return S02_T(k1_til_sq, k2_til_pm_sq, kR_sq, eps_pm * one_tilde**2)

def S11_R(k1_til_sq, k2_til_pm_sq, kR_sq, eps_pm, one_tilde):    
    return S02_R(k1_til_sq, k2_til_pm_sq, kR_sq, eps_pm * one_tilde**2)
```

### $\mathcal{S}_{0,2}(\omega)$ calculations

\begin{align*}
    \mathcal{S}_{0,2} &= 4 \Delta \chi^2 \epsilon_0^3 \hbar \omega^6 \Theta(\omega) N(\omega) \Im[\epsilon_{\textrm{eq}}(\omega)] \sum_{n=\pm 1} \omega_n^2 \mu_0^4 \mathcal{G}^n_{0,2} \\
    \mathcal{G}^n_{0,2}(\vec{r}, \omega) &= \Re \operatorname{Tr} \int_{\rho_z < 0} d^3 \vec{\rho} \int_{\rho_z' < 0} d^3 \vec{\rho}' \int_{\rho_z'' < 0} d^3 \vec{\rho}'' G(\omega, \vec{r}, \vec{\rho}) G(\omega_n, \vec{\rho}, \vec{\rho}') G(\omega, \vec{\rho}', \vec{\rho}'') G(\omega, \vec{r}, \vec{\rho}'')^\dagger 
\end{align*}

- Incandescent: Eq. (S.VI.29), differ by factor of $\pi$. 

```python
def S02(w, z, Omega, amp): # returns real number
    w_pos = w + Omega
    w_neg = w - Omega
    
    G02_pos = bbarG(w, z, Omega, +1)
    G02_neg = bbarG(w, z, Omega, -1)
    
    eps = permittivity(w)
    
    res = w_pos**2 * G02_pos + w_neg**2 * G02_neg
    res *= 4 * amp**2 * w**6 / (eps0 * c_speed**8) * eps.imag * bose(w)
    
    return res
```

#### Computing the Green functions

- Incandescent: Eq. (S.VI.58)

\begin{align*}
    \mathcal{G}^{\pm}_{0,2} = \bar{\bar{\mathcal{G}}}^{(\omega^\pm)}_{TT (0,2)} 
                                                + \bar{\bar{\mathcal{G}}}^{(\omega^\pm)}_{TR (0,2)} 
                                                + \bar{\bar{\mathcal{G}}}^{(\omega^\pm)}_{TZ (0,2)} 
                                                + \bar{\bar{\mathcal{G}}}^{(\omega^\pm)}_{RT (0,2)} 
                                                + \bar{\bar{\mathcal{G}}}^{(\omega^\pm)}_{RR (0,2)} 
                                                + \bar{\bar{\mathcal{G}}}^{(\omega^\pm)}_{RZ (0,2)}
                                                + \bar{\bar{\mathcal{G}}}^{(\omega^\pm)}_{ZT (0,2)} 
                                                + \bar{\bar{\mathcal{G}}}^{(\omega^\pm)}_{ZR (0,2)} 
                                                + \bar{\bar{\mathcal{G}}}^{(\omega^\pm)}_{ZZ (0,2)}
\end{align*}

Since the integrals are similar, we add the integrands first before integrating.

```python
# leave out common factor of 1/(32 pi k0^5)
def bbarG_integrand(params):
    return (
          bbarGTT_integrand(params)
        + bbarGTR_integrand(params)
        + bbarGTZ_integrand(params)
        + bbarGRT_integrand(params)
        + bbarGRR_integrand(params)
        + bbarGRZ_integrand(params)
        + bbarGZT_integrand(params)
        + bbarGZR_integrand(params)
        + bbarGZZ_integrand(params)
    )

def bbarG_integrand_wrapper(kR, w, z, Omega, sign):
    params = get_params(kR, w, z, Omega=Omega, sign=sign)
    return bbarG_integrand(params)

# plot_integrand("bbarG", z, Omega, +1)
# plot_integrand("bbarG", z, Omega, -1)
```

```python
def bbarG(w, z, Omega, sign):
    left_end = 0
    # right_end = 4
    right_end = np.infty
    
    integral, _ = integrate.quad(bbarG_integrand_wrapper, 
                                 left_end, 
                                 right_end, 
                                 args=(w, z, Omega, sign))
    
    k0 = w / c_speed
    return integral / (32 * pi * k0**5)
```

#### $\bar{\bar{\mathcal{G}}}^{(\omega^\pm)}_{TT (0,2)}$ calculations

<font color='blue'>
    Supp page 28 (S.VI.59)
</font>

\begin{align*}
    \bar{\bar{\mathcal{G}}}^{(\omega^\pm)}_{TT (0,2)} = \frac{1}{32 \pi k_0^5} \int_{0}^{\infty} d \kappa_R \,
        \frac{-\kappa_R}{|\tilde{k}_{2}|^2 \tilde{k}_{2} \tilde{k}_{2}^{\pm}}
        \sum_{i,j \in \{\uparrow, \downarrow\}} \bar{\bar{\mathcal{G}}}^{\pm (ij)}_{TT (0,2)} \bar{\bar{\Gamma}}^{\pm (ij)}_{TT (0,2)}
\end{align*}

(For the code below, integrand means without the coefficients outside the integral.)

```python
def bbarGTT_integrand(params):
    Gamma_TT_00 = bbarGammaTT00(params)
    Gamma_TT_01 = bbarGammaTT01(params)
    Gamma_TT_10 = bbarGammaTT10(params)
    Gamma_TT_11 = bbarGammaTT11(params)
    
    G_TT_00 = bbarGTTij(0, 0, params)
    G_TT_01 = bbarGTTij(0, 1, params)
    G_TT_10 = bbarGTTij(1, 0, params)
    G_TT_11 = bbarGTTij(1, 1, params)
    
    res = G_TT_00 * Gamma_TT_00 + G_TT_01 * Gamma_TT_01 + G_TT_10 * Gamma_TT_10 + G_TT_11 * Gamma_TT_11
    res *= -params['kR'] / (abs(params['k2_til'])**2 * params['k2_til'] * params['k2_til_pm'])
    
    return res.real
```

<font color='blue'>
    Supp page 28-29
</font>

\begin{align*}
    \bar{\bar{\mathcal{G}}}^{\pm (\uparrow\uparrow)}_{TT (0,2)} &= |t^s|^2 + |t^p|^2 \frac{T}{(1 \pm \tilde{\Omega})^2 \epsilon(\omega) \epsilon(\omega^\pm)} [\tilde{k}_{2} \tilde{k}_2^\pm + \kappa_R^2] [\tilde{k}_{2} \tilde{k}_2^\pm + \kappa_R^2] \\
    \bar{\bar{\mathcal{G}}}^{\pm (\uparrow\downarrow)}_{TT (0,2)} &= |t^s|^2 + |t^p|^2 \frac{R}{(1 \pm \tilde{\Omega})^2 \epsilon(\omega) \epsilon(\omega^\pm)} [\tilde{k}_{2} \tilde{k}_2^\pm + \kappa_R^2] [\tilde{k}_{2} \tilde{k}_2^\pm - \kappa_R^2] \\
    \bar{\bar{\mathcal{G}}}^{\pm (\downarrow\uparrow)}_{TT (0,2)} &= |t^s|^2 + |t^p|^2 \frac{T}{(1 \pm \tilde{\Omega})^2 \epsilon(\omega) \epsilon(\omega^\pm)} [\tilde{k}_{2} \tilde{k}_2^\pm - \kappa_R^2] [\tilde{k}_{2} \tilde{k}_2^\pm - \kappa_R^2] \\
    \bar{\bar{\mathcal{G}}}^{\pm (\downarrow\downarrow)}_{TT (0,2)} &= |t^s|^2 + |t^p|^2 \frac{R}{(1 \pm \tilde{\Omega})^2 \epsilon(\omega) \epsilon(\omega^\pm)} [\tilde{k}_{2} \tilde{k}_2^\pm - \kappa_R^2] [\tilde{k}_{2} \tilde{k}_2^\pm + \kappa_R^2] 
\end{align*} 

Programming notes:
- Combine all four functions in a single definition, using two indices `i, j`, taking the possible values of $\uparrow = 0$, $\downarrow = 1$.
- To dynamically determine the sign preceding the two $\kappa_R$:
    1. $(-1)^{i}$
    2. $(-1)^{i+j}$

```python
def bbarGTTij(i, j, params):
    sign1 = (-1)**i
    sign2 = (-1)**(i+j)
    
    if j == 0:
        TR_factor = params['T'] / params['denom']
    else: # j ==1
        TR_factor = params['R'] / params['denom']
    
    return (
        params['ts_sq'] + params['tp_sq'] * TR_factor 
        * (params['k2k2'] + sign1 * params['kR_sq']) 
        * (params['k2k2'] + sign2 * params['kR_sq'])
    )
```

<font color='blue'>
    Supp page 28-29
</font>

\begin{align*}
    \bar{\bar{\Gamma}}^{\pm (\uparrow\uparrow)}_{TT} &= \frac{e^{-2 k_0 \Im(\tilde{k}_1) z}}
    {4i [ \Im(\tilde{k}_2) ]^2 [\tilde{k}_2^* - \tilde{k}_2^{\pm}]} \\
    \bar{\bar{\Gamma}}^{\pm (\uparrow\downarrow)}_{TT} &= \frac{e^{-2 k_0 \Im(\tilde{k}_1) z}}
    {\tilde{k}_2 - \tilde{k}_2^{\pm}} 
    \biggl[ \frac{i}{[\tilde{k}_2 + \tilde{k}_2^{\pm}][\tilde{k}_2^* - \tilde{k}_2^{\pm}]} 
    + \frac{1}{4 \tilde{k}_2 \Im(\tilde{k}_2)} \biggr] \\
    \bar{\bar{\Gamma}}^{\pm (\downarrow\uparrow)}_{TT} &= \frac{e^{-2 k_0 \Im(\tilde{k}_1) z}}
    {-4i [\Im(\tilde{k}_2)]^2 [\tilde{k}_2 + \tilde{k}_2^{\pm}]} \\
    \bar{\bar{\Gamma}}^{\pm (\downarrow\downarrow)}_{TT} &= \frac{e^{-2 k_0 \Im(\tilde{k}_1) z}}
    {-4 [\tilde{k}_2 + \tilde{k}_2^{\pm}] \tilde{k}_2 \Im(\tilde{k}_2)}
\end{align*} 

Note: 
- when $\kappa_R \leq 1$, $\Im(\tilde{k}_1) = 0$

```python
def bbarGammaTT00(params):
    return (
        params['decay'] / (
            4j * params['k2_til'].imag**2
            * (params['k2_til'].conjugate() - params['k2_til_pm'])
        ) 
    )
```

```python
def bbarGammaTT01(params):
    return (
        params['decay'] / (params['k2_til'] - params['k2_til_pm']) 
        * (
            1j / (
                (params['k2_til'] + params['k2_til_pm']) 
                * (params['k2_til'].conjugate() - params['k2_til_pm'])
            ) 
            + 1/(4 * params['k2_til'] * params['k2_til'].imag)
        )
    )
```

```python
def bbarGammaTT10(params):
    return (
        params['decay'] / (
            -4j * params['k2_til'].imag**2
            * (params['k2_til'] + params['k2_til_pm']) 
        )
    )
```

```python
def bbarGammaTT11(params):
    return (
        params['decay'] / (
            -4 * (params['k2_til'] + params['k2_til_pm'])
            * params['k2_til'] * params['k2_til'].imag
        )
    )
```

#### $\bar{\bar{\mathcal{G}}}^{(\omega^\pm)}_{R-R (0,2)}$ calculations

        \bar{\bar{\mathcal{G}}}^{\pm (\downarrow \uparrow)}_{RR (0,2)} \bar{\bar{\Gamma}}^{\pm}_{RR (0,2)}

```python
def bbarGRR_integrand(params):
    Gamma_RR = bbarGammaRR(params)
    
    G_RR_10 = bbarGRR10(params)
    
    res = G_RR_10 * Gamma_RR
    res *= -params['kR']/(abs(params['k2_til'])**2 * params['k2_til'] * params['k2_til_pm'])
    
    return res.real
```

<font color='blue'>

\begin{align*}
    \bar{\bar{\mathcal{G}}}^{\pm (\downarrow \uparrow)}_{RR (0,2)} = |t^s|^2 r^{s(\pm)} r^s
        + |t^p|^2 r^{p(\pm)} r^p
        \frac{T}{(1 \pm \tilde{\Omega})^2 \epsilon(\omega) \epsilon(\omega^\pm)} [\tilde{k}_{2} \tilde{k}_2^\pm - \kappa_R^2]^2
\end{align*}

```python
def bbarGRR10(params):
    TR_factor = params['T'] / params['denom']
    
    return (
        params['ts_sq'] * params['rs'] * params['rs_pm']
        + params['tp_sq'] * params['rp'] * params['rp_pm']
        * TR_factor * (params['k2k2'] - params['kR_sq'])**2
    )
```

\begin{align*}
    \bar{\bar{\Gamma}}^{\pm}_{RR (0,2)} = \frac{e^{-2 k_0 \Im(\tilde{k}_1) z}}
    {-2 \Im(\tilde{k}_2) [\tilde{k}_2 + \tilde{k}_2^{\pm}]^2}
\end{align*}

```python
def bbarGammaRR(params):
    return params['decay']/(
        -2 * params['k2_til'].imag * (params['k2_til'] + params['k2_til_pm'])**2
    )
```

#### $\bar{\bar{\mathcal{G}}}^{(\omega^\pm)}_{T-R (0,2)}$ calculations

<font color='blue'>
    Supp page 29 (S.VI.62)
</font>

\begin{align*}
    \bar{\bar{\mathcal{G}}}^{(\omega^\pm)}_{TR (0,2)} = \frac{1}{32 \pi k_0^5} \int_{0}^{\infty} d \kappa_R \,
        \frac{-\kappa_R}{|\tilde{k}_{2}|^2 \tilde{k}_{2} \tilde{k}_{2}^{\pm}}
        \sum_{i \in \{\uparrow, \downarrow\} } \bar{\bar{\mathcal{G}}}^{\pm (i \uparrow)}_{TR (0,2)} \bar{\bar{\Gamma}}^{\pm (i)}_{TR (0,2)}
\end{align*}

```python
def bbarGTR_integrand(params):
    Gamma_TR = bbarGammaTR(params)
    
    G_TR = bbarGTR(params)
    
    res = 2 * G_TR * Gamma_TR
    res *= -params['kR']/(abs(params['k2_til'])**2 * params['k2_til'] * params['k2_til_pm'])
    
    return res.real
```

<font color='blue'>
    Supp page 29
</font>

\begin{align*}
    \bar{\bar{\mathcal{G}}}^{\pm (\uparrow \uparrow)}_{TR (0,2)} &= |t^s|^2 r^s + |t^p|^2 r^p \frac{T}{(1 \pm \tilde{\Omega})^2 \epsilon(\omega) \epsilon(\omega^\pm)} [\tilde{k}_{2} \tilde{k}_2^\pm + \kappa_R^2] [\tilde{k}_{2} \tilde{k}_2^\pm - \kappa_R^2] = \bar{\bar{\mathcal{G}}}^{\pm (\downarrow \uparrow)}_{TR (0,2)}

```python
def barGTTij(i, j, params):
    sign1 = (-1)**i
    sign2 = (-1)**j
    
    if j == 0:
        TR_factor = params['T_pm'] / abs(params['denom'])
    else: # j ==1
        TR_factor = params['R_pm'] / abs(params['denom'])
    
    return (
        params['ts_sq'] + params['tp_sq'] * TR_factor 
        * (params['k2k2'] + sign1 * params['kR_sq']).conjugate() 
        * (params['k2k2'] + sign2 * params['kR_sq'])
    )
```

```python
def barGammaTT00(params):
    return (
        2 * params['decay']
        / (abs(params['k2_til'] - params['k2_til_pm'])**2) 
        * (params['k2_til'] + params['k2_til_pm']).imag 
        * (
            1/(4 * params['k2_til'].imag * params['k2_til_pm'].imag) 
            - 1/abs(params['k2_til'] - params['k2_til_pm'].conjugate())**2
        )
    )
```

```python
def barGTR_integrand(params):
    Gamma_TR_01 = barGammaTR01(params)
    Gamma_TR_11 = barGammaTR11(params)
    
    G_TR_01 = barGTRi(0, params)
    G_TR_11 = barGTRi(1, params)
    
    res = G_TR_01 * Gamma_TR_01 + G_TR_11 * Gamma_TR_11
    res *= params['kR']/(abs(params['k2_til'])**2 * abs(params['k2_til_pm'])**2)
    
    return res.real
```
